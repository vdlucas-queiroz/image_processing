### Aula 4 - NumPy

NumPy é uma biblioteca open source que permite diferentes cálculos numéricos em Python. Ele fornece suporte para trabalhar com arrays multidimensionais (chamados de `ndarray`) e uma variedade de funções matemáticas, lógicas e estatísticas para realizar operações de alto desempenho sobre esses arrays.

A documentação do numpy pode ser encontrada em https://numpy.org/doc/stable/ .

Vantagens do Numpy :

- O objeto principal do NumPy é o ndarray, que é uma estrutura eficiente e flexível para armazenar e manipular grandes quantidades de dados numéricos;

- O NumPy permite realizar operações em arrays inteiros (vetorização) sem a necessidade de loops explícitos, tornando o código mais limpo e eficiente;

- Foi implementado em C, garantindo alto desempenho em comparação com listas tradicionais do Python;

- É amplamente utilizado em ecossistemas de computação científica e aprendizado de máquina, sendo a base para bibliotecas como Pandas, SciPy, Matplotlib e Scikit-learn;

- Permite operações como indexação avançada, slicing, reshaping (mudança de formato), e broadcasting (operações automáticas entre arrays de diferentes formas).

#### Exemplo:

A equação da densidade de probabilidade para a Distribuição Gaussiana Multidimensional é:

$$
f(\mathbf{x}) = \frac{1}{\sqrt{(2\pi)^N |\Sigma|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu})^T \Sigma^{-1} (\mathbf{x} - \boldsymbol{\mu}) \right)
$$

```python
# Exemplos (CHATGPT)

# Parâmetros da distribuição
mu = np.array([1, 2])  # Vetor de médias
cov = np.array([[1, 0.5], [0.5, 2]])  # Matriz de covariância (2x2)

# Ponto onde queremos calcular a probabilidade
x = np.array([1.5, 2.5])
```

```python
len(mu)
```

```python
# Dimensão
k = len(mu)

# Determinante e inversa da matriz de covariância
det_cov = np.linalg.det(cov)
inv_cov = np.linalg.inv(cov)

# Constante do termo normalizador
normalizer = 1 / (np.sqrt((2 * np.pi)**k * det_cov))

# Termo exponencial
diff = x - mu
exponent = -0.5 * np.dot(np.dot(diff.T, inv_cov), diff,)

# Densidade de probabilidade
probability = normalizer * np.exp(exponent)
probability
```

```python
# Dados de exemplo
mu = np.array([1, 2])  # Vetor de médias
cov = np.array([[2, 0.5], [0.5, 1]])  # Matriz de covariância
inv_cov = np.linalg.inv(cov)  # Inversa da matriz de covariância

# Pontos a serem avaliados (m pontos no espaço de dimensão k)
X = np.array([
    [1.5, 2.5],  # Ponto 1
    [0.0, 1.0],  # Ponto 2
    [2.0, 3.0]   # Ponto 3
])

# Passo 1: Diferença entre os pontos e a média
diff = X - mu  # Matriz diff de forma (m, k)

# Cálculo 1: Forma escalar por ponto
# Para cada ponto, calculamos d^T * inv_cov * d de forma direta
results_manual = []
for d in diff:
    value = d.T @ inv_cov @ d  # Produto vetorial completo
    results_manual.append(value)
results_manual = np.array(results_manual)

# Cálculo 2: Forma vetorizada com broadcasting
diff_transformed = diff @ inv_cov  # Produto vetorial (m, k) @ (k, k) -> (m, k)
results_vectorized = np.sum(diff_transformed * diff, axis=1)  # Multiplicação elemento a elemento e soma por linha

# Comparação dos resultados
print("Resultados manual (produto vetorial completo):", results_manual)
print("Resultados vetorizados (produto + elemento a elemento):", results_vectorized)

# Verificação da igualdade
assert np.allclose(results_manual, results_vectorized), "Os resultados não são idênticos!"
print("Os resultados são idênticos.")
```

#### Exemplo de interoperabilidade com o pandas

```python
# Importando pandas
import pandas as pd
```

```python
# Simulação de um DataFrame com 5 classes e valores para 3 bandas
np.random.seed(42)
data = {
    "class": ["Vegetation", "Water", "Urban", "Soil", "Vegetation", 
              "Urban", "Water", "Soil", "Vegetation", "Urban"],
    "band1": np.random.uniform(0, 1, 10),  # Valores simulados para banda 1
    "band2": np.random.uniform(0, 1, 10),  # Valores simulados para banda 2
    "band3": np.random.uniform(0, 1, 10)   # Valores simulados para banda 3
}
df = pd.DataFrame(data)
df
```

```python

df["class"] = df["class"].astype("category")  # Convertendo a coluna 'class' para categórica

df
```

```python
df['class'].dtype
```

```python
df['class'].cat.categories
```

```python
# Função para calcular vetor de médias e matriz de covariância
def calculate_stats(df, class_name):
    # Filtrar os dados para a classe específica
    class_data = df[df["class"] == class_name][["band1", "band2", "band3"]].values

    # Vetor de médias
    mean_vector = np.mean(class_data, axis=0)

    # Matriz de covariância
    cov_matrix = np.cov(class_data, rowvar=False)  # rowvar=False: colunas são variáveis
    
    return mean_vector, cov_matrix
```

```python
class_stats = {} #dicionário para armazernar os 
```

```python
# Calcular para cada classe
for cls in df["class"].cat.categories:
    mean_vector, cov_matrix = calculate_stats(df, cls)
    class_stats[cls] = {"mean_vector": mean_vector, "cov_matrix": cov_matrix}
```
