4. Atualização de variáveis e Aliasing 

````python
# Função id() retorna o identificador (geralmente o endereço de memória)

a = [1, 2, 3]
b = [2, 3, 4]

print(id(a))
print(id(b))
print(id(a) == id(b))
````

````python
# Quando os objetos são mutáveis teremos problemas de 'confusão de referência' também chamado de 'aliasing' 

a = [1, 2, 3]
c = [1, 2, 3]  # 'c' e 'a' referenciam a mesma lista
print(c)

print(f'Endereço de memória de a: {id(a)}.\nEndereço de memória de c: {id(c)}.')
print(id(a)==id(c)) # comparando valores
````

````python
# Comparando com o operador 'is' - Compara se os dois objetos são o mesmo!

c is a
````

````python
# Alterando o valor de uma lista com o método append() e verificando aliasing

c.append(4)  # Modificando a lista através de 'c'
print(a)  # Saída: [1, 2, 3, 4] (também foi alterada)
````

#### Funções e métodos

* Uma função é um bloco de código reutilizável que realiza uma tarefa específica. Ela pode receber argumentos (dados de entrada), processá-los e, opcionalmente, retornar um resultado. (GENERALISTA)
* Um método é semelhante a uma função, mas está associado a um objeto (ou seja, é uma função pertencente a uma classe). Métodos atuam sobre os dados do objeto ao qual estão associados. (ESPECIFISTA)

````python
# Para criar uma função, utilizamos o comando def, damos o nome da função e identamos o código. 
# Num código corrido, a função pode vir em qualquer lugar do código. No caso de notebooks, como as células são independentes no que se refere à execução, 
# elas devem ser rodadas antes ou simultaneamente com o código onde está sua chamada.

# No R as funções são criadas dessa forma:

""" # Função que calcula a soma de dois números
soma <- function(a, b) {
  return(a + b)
}

# Chamando a função
resultado <- soma(5, 3)
cat("Resultado em R:", resultado, "\n")  # Saída: Resultado em R: 8 """

# No Python, nós precisamento primeiro definir a função antes de associar ela com uma variável e não precisamentos da palavra reservada 'function'. 
# Além disso, como vimos, é necessário identação nos blocos internos.

def soma(a,b):
    return(a + b)


# As variáveis utilizadas dentro da função não estão no escopo global.
````

````python
# Usando a função soma

a = 10
b = 15

soma(a,b)
````

````python
# Exemplo um pouco fora do escopo...

class Calculadora:
    def __init__(self):
        pass
    
    def soma(self, a, b):
        return a + b
````
